```python
def assign_score_range(score):
    if pd.isna(score):
        return None

    try:
        score = float(score)
    except (TypeError, ValueError):
        return None

    if 70 <= score <= 80:
        return "80-70"
    elif 60 <= score < 70:
        return "69-60"
    elif 50 <= score < 60:
        return "59-50"
    elif 40 <= score < 50:
        return "49-40"
    elif 30 <= score < 40:
        return "39-30"
    elif 20 <= score < 30:
        return "29-20"
    elif 0 <= score < 20:
        return "19-0"

    return None


score_range_order = [
    "80-70",
    "69-60",
    "59-50",
    "49-40",
    "39-30",
    "29-20",
    "19-0"
]

outcome_order = [
    "Quick Match",
    "Manual Match",
    "Create New",
    "Not Mapped"
]
```

```python
test_scores = [80, 75, 69, 60, 55, 40, 39, 20, 19, 0, 100]

for score in test_scores:
    print(score, assign_score_range(score))
```

```python
data = test_data.copy()

data.columns = ["Match Score", "Final Mapping Status"]

data["Match Score"] = pd.to_numeric(
    data["Match Score"],
    errors="coerce"
)

data["Final Mapping Status"] = (
    data["Final Mapping Status"]
    .astype("string")
    .str.strip()
)

data = data[
    data["Final Mapping Status"].notna()
    & (data["Final Mapping Status"] != "")
].copy()

data = data[
    data["Final Mapping Status"].isin(outcome_order)
].copy()

data["Score Range"] = data["Match Score"].apply(assign_score_range)

data = data[data["Score Range"].notna()].copy()

data.head()
```

```python
summary_rows = []

for workbook in workbooks:

    input_file = workbook["path"]
    folder_category = workbook["folder"]

    workbook_name = input_file.stem

    if workbook_name[:8].isdigit():
        workbook_name = workbook_name[8:].strip()

    workbook_name = workbook_name.replace("Match Review", "").strip()

    for worksheet_name in worksheet_names:

        data = pd.read_excel(
            input_file,
            sheet_name=worksheet_name,
            usecols="W,BR"
        )

        data.columns = ["Match Score", "Final Mapping Status"]

        data["Match Score"] = pd.to_numeric(
            data["Match Score"],
            errors="coerce"
        )

        data["Final Mapping Status"] = (
            data["Final Mapping Status"]
            .astype("string")
            .str.strip()
        )

        data = data[
            data["Final Mapping Status"].notna()
            & (data["Final Mapping Status"] != "")
        ].copy()

        data = data[
            data["Final Mapping Status"].isin(outcome_order)
        ].copy()

        data["Score Range"] = data["Match Score"].apply(assign_score_range)

        data = data[data["Score Range"].notna()].copy()

        counts = pd.crosstab(
            data["Score Range"],
            data["Final Mapping Status"]
        )

        counts = counts.reindex(
            index=score_range_order,
            columns=outcome_order,
            fill_value=0
        )

        for score_range in score_range_order:

            quick_match = int(counts.loc[score_range, "Quick Match"])
            manual_match = int(counts.loc[score_range, "Manual Match"])
            create_new = int(counts.loc[score_range, "Create New"])
            not_mapped = int(counts.loc[score_range, "Not Mapped"])

            total_records = (
                quick_match
                + manual_match
                + create_new
                + not_mapped
            )

            summary_rows.append({
                "Folder": folder_category,
                "Workbook": workbook_name,
                "Worksheet": worksheet_name,
                "Score Range": score_range,
                "Quick Match": quick_match,
                "Manual Match": manual_match,
                "Create New": create_new,
                "Not Mapped": not_mapped,
                "Total Records": total_records
            })

print("Finished reading all workbooks and worksheets.")
```

```python
sheet_name = "Worksheet Score Analysis"

with pd.ExcelWriter(
    output_file,
    engine="xlsxwriter"
) as writer:

    results_with_total.to_excel(
        writer,
        sheet_name=sheet_name,
        index=False
    )

    workbook = writer.book
    worksheet = writer.sheets[sheet_name]

    header_format = workbook.add_format({
        "bold": True,
        "text_wrap": True,
        "align": "center",
        "valign": "vcenter"
    })

    body_format = workbook.add_format({
        "valign": "vcenter"
    })

    center_format = workbook.add_format({
        "align": "center",
        "valign": "vcenter"
    })

    for column_number, column_name in enumerate(results_with_total.columns):
        worksheet.write(
            0,
            column_number,
            column_name,
            header_format
        )

    worksheet.set_column("A:A", 27, body_format)
    worksheet.set_column("B:B", 25, body_format)
    worksheet.set_column("C:C", 14, center_format)
    worksheet.set_column("D:D", 13, center_format)
    worksheet.set_column("E:I", 15, center_format)

    worksheet.set_row(0, 32)

    worksheet.freeze_panes(1, 0)

    worksheet.autofilter(
        0,
        0,
        len(results_with_total),
        len(results_with_total.columns) - 1
    )

print("Analysis complete.")
print(output_file)
```
